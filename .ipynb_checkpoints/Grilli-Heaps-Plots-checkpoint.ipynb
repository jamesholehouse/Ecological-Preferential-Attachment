Need to automate this procedure for all categories.

## Automating the plots.

```julia
heaps_plots = [];
exponents = ones(0);
for j in 1:length(gd_classes)
    # first need the exponents calculated from GLM.
    logxs = log10.(sum.([x[!,:count] for x in gd_classes[j]]));
    logys = log10.(length.([x[!,:count] for x in gd_classes[j]]));
    df = DataFrame(y = logys, x = logxs);
    glm = lm(@formula(y ~ x), df);
    push!(exponents, coef(glm)[2])
    
    # make the plot
    xvals = log10.(sum.([x[!,:count] for x in gd_classes[j]]));
    yvals = log10.(length.([x[!,:count] for x in gd_classes[j]]));
    heap_plot = scatter(xvals, yvals, alpha = 0.5, label = :none, c = pal[8]);
    plot!(collect(0:7),coef(glm)[1] .+ collect(0:7) .*coef(glm)[2], label = L"x^{%$(round(coef(glm)[2],digits = 2))}", color = :black, lw = 2)
    plot!(xlims = [minimum(xvals)-0.5,maximum(xvals)+0.5], ylims = [minimum(yvals)-0.5,maximum(yvals)+0.5], legend = :topleft,
        size = (400,300), grid = :none, minorgrid = :none, legendfontsize = 12, title = classes[j])
    # ylabel = L"$\log_{10}$(unique species)", xlabel = L"$\log_{10}$(species abundance)"
    if j == 6
        plot!(ylabel = L"$\log_{10}$(unique species)", ylabelfontsize = 15)
    elseif j == 13
        plot!(xlabel = L"$\log_{10}$(species abundance)", xlabelfontsize = 15)
    end
    push!(heaps_plots, heap_plot)
end    
```
